```python
import pandas as pd
import numpy as np

data = {
    'Customer_ID': [101, 102, 103, 104, 105, 105, 107, 108, 109, 110],
    'Name': [' Anish ', ' Rahul', 'Sreeja', 'Devanand', ' Maya ', ' Maya ', 'Arun', ' Kavya', 'Nikhil', 'Priya'],
    'Age': [25, np.nan, 30, 22, 28, 28, 150, 24, np.nan, 29], # 150 എന്നത് ഒരു Outlier ആണ്
    'Gender': ['Male', 'male', 'F', 'M', 'Female', 'Female', 'M', 'Female', 'male', 'F'],
    'Joining_Date': ['2023-01-15', '15/01/2023', '2023-02-20', '2023-03-10', '2023-03-12', '2023-03-12', '2023/04/01', np.nan, '2023-05-18', '2023-06-22'],
    'Salary': ['50000', '60000', ' 55000 ', '70000', '65000', '65000', '80000', np.nan, '48000', '75000']
}

df_messy = pd.DataFrame(data)
df_messy.to_csv("messy_dataset.csv", index=False)
print("Messy Dataset Successfully Created!")
```

```python
# Load messy data
df = pd.read_csv("messy_dataset.csv")
print("2. Initial Messy Data:")
df
```

```python
print("--- Data Quality Report ---")
print("1. Total Rows & Columns:", df.shape)
print("\n2. Null Values Count:\n", df.isnull().sum())
print("\n3. Duplicate Rows Count:", df.duplicated().sum())
print("\n4. Data Types:\n", df.dtypes)
```

```python

df_clean = df.drop_duplicates().copy()

df_clean['Name'] = df_clean['Name'].astype(str).str.strip()

df_clean['Gender'] = df_clean['Gender'].astype(str).str.strip().str.upper()
df_clean['Gender'] = df_clean['Gender'].replace({'M': 'Male', 'MALE': 'Male', 'F': 'Female', 'FEMALE': 'Female'})

df_clean['Salary'] = pd.to_numeric(df_clean['Salary'].astype(str).str.strip(), errors='coerce')

df_clean['Salary'] = df_clean['Salary'].fillna(df_clean['Salary'].median())
df_clean['Age'] = pd.to_numeric(df_clean['Age'], errors='coerce')
df_clean['Age'] = df_clean['Age'].fillna(df_clean['Age'].median())

df_clean.loc[df_clean['Age'] > 100, 'Age'] = df_clean['Age'].median()

df_clean['Joining_Date'] = pd.to_datetime(df_clean['Joining_Date'], errors='coerce')
df_clean['Joining_Date'] = df_clean['Joining_Date'].ffill()

print("Cleaned Dataset:")
df_clean
```

```python
summary_data = {
    'Metric': ['Total Rows', 'Duplicate Rows', 'Null Values in Age', 'Null Values in Salary'],
    'Before Cleaning': [len(df), df.duplicated().sum(), df['Age'].isnull().sum(), df['Salary'].isnull().sum()],
    'After Cleaning': [len(df_clean), df_clean.duplicated().sum(), df_clean['Age'].isnull().sum(), df_clean['Salary'].isnull().sum()]
}

summary_df = pd.DataFrame(summary_data)
print("--- Before vs After Summary ---")
print(summary_df)

df_clean.to_csv("cleaned_dataset.csv", index=False)
print("\nCleaned dataset saved as 'cleaned_dataset.csv'!")
```
